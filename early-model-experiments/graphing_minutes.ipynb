```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd

df = pd.read_csv('game_logs/moussa_cisse_game_log.csv')

def mp_to_minutes(x):
    if pd.isna(x):
        return 0

    x = str(x).strip()

    if x.lower() in ["did not play", "inactive"]:
        return 0

    if ":" in x:
        m, s = map(int, x.split(":"))
        return m  # Ignore seconds (or use m + s/60 to keep them)

    return float(x)

df['MP'] = df['MP'].apply(mp_to_minutes)

df['Date'] = pd.to_datetime(df['Date'])
df = df.sort_values('Date')

plt.figure(figsize=(10, 5))
plt.bar(df['Date'], df['MP'])

plt.xlabel('Date')
plt.ylabel('Minutes Played')
plt.title('Moussa Cisse Minutes Played vs Date')

# Show 0–48 on the y-axis
plt.ylim(0, 48)
plt.yticks(range(0, 49, 6))  # 0, 6, 12, ..., 48

plt.xticks(rotation=45, ha='right')
plt.tight_layout()

# Format the x-axis to show one tick per month
ax = plt.gca()
ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b'))

# Limit the season from October to April
plt.xlim(pd.Timestamp('2025-10-01'), pd.Timestamp('2026-04-30'))
plt.show()
```
